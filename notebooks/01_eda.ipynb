# ISMPCE — Exploratory Data Analysis & Feature Engineering
## Case Study: CO2 Emission Monitoring & Forecasting (Bauchi State / Nigeria)

This notebook conducts an in-depth exploratory data analysis (EDA) of Nigeria's annual CO2 emissions (1990–2024), combining datasets from:
1. **Our World in Data (OWID)** — historical CO2, GDP, population, fuel shares
2. **World Bank (WDI)** — macroeconomic indicators and energy access
3. **EDGAR v8.0** — sectoral breakdown (Energy, Transport/Oil, Flaring, Industrial)

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
```

### 1. Load Processed Dataset

```python
data_path = os.path.join('..', 'data', 'processed', 'merged_emissions.csv')
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape}")
df.head()
```

### 2. Historical CO2 Emissions Trend (1990–2024)

```python
plt.figure(figsize=(14, 5))
plt.plot(df['year'], df['co2'], marker='o', color='#2563eb', linewidth=2.5, label='Total CO2 (Mt)')
plt.title('Nigeria Historical CO2 Emissions (1990 - 2024)', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Emissions (Million Metric Tonnes)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

### 3. Sectoral Breakdown (Energy, Transport/Oil, Flaring, Industrial)

```python
sector_cols = ['energy_co2', 'oil_transport_co2', 'flaring_co2', 'industrial_co2']
labels = ['Energy & Power', 'Oil & Transport', 'Gas Flaring', 'Industrial & Cement']
colors = ['#3b82f6', '#f97316', '#ef4444', '#8b5cf6']

plt.figure(figsize=(14, 6))
plt.stackplot(df['year'], [df[col] for col in sector_cols], labels=labels, colors=colors, alpha=0.85)
plt.title('CO2 Emissions by Economic Sector (1990 - 2024)', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('CO2 Emissions (Mt)')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

### 4. Correlation Matrix of Key Macroeconomic and Emission Features

```python
corr_features = ['co2', 'gdp', 'population', 'primary_energy_consumption', 'energy_per_capita', 'electricity_access_pct', 'oil_transport_co2', 'flaring_co2']
corr = df[corr_features].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', square=True, linewidths=0.5)
plt.title('Correlation Heatmap of Key Indicators', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5. Feature Engineering Summary
- **Lag features**: $CO_2(t-1), CO_2(t-2), CO_2(t-3)$
- **Rolling averages**: 3-year and 5-year rolling windows
- **Fossil fuel share**: $\frac{Coal + Oil + Gas}{Total CO_2}$
- **Interaction terms**: $GDP \times Population$, Energy Intensity
- **Normalization**: Min-Max scaling $[0, 1]$ applied for sequence / neural networks.
